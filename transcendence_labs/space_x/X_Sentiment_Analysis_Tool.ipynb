```python
import tweepy
from textblob import TextBlob
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# credentials
consumer_key = 'wKiWEjagqVdAQo4sKvlnutWLr'
consumer_secret = 'gvKcI2hTyRleUJbUDsCLHKEoBsGh9mAdrgz0VIzTNTkw5LtiA3'
access_token = '1783773742294011904-EpvDTDpUcwmGWDtHI8JkBvROW5W1PI'
access_token_secret = 'HemyxIn3mGjcQTFhm824V1894q0u3OLOrppA0mNT1uHdW'

# Authenticate
auth = tweepy.OAuth1UserHandler(consumer_key, consumer_secret, access_token, access_token_secret)
api = tweepy.API(auth)

# Specify the username (without @)
username = 'The_REDWRITER'

# Fetch the last 10 posts
tweets = api.user_timeline(screen_name=username, count=10)

replies = []
for tweet in tweets:
    # Search for replies directed to the user in the context of the post's conversation
    query = f"to:{username} conversation_id:{tweet.id}"
    for reply in tweepy.Cursor(api.search_tweets, q=query).items():
        # Filter for direct replies to this specific post
        if reply.in_reply_to_status_id == tweet.id:
            replies.append(reply)

sentiment_scores = []
for reply in replies:
    # Analyze sentiment
    analysis = TextBlob(reply.full_text)
    polarity = analysis.sentiment.polarity  # Ranges from -1 (negative) to 1 (positive)
    
    # Convert polarity to a percentage (0% = extremely negative, 100% = extremely positive)
    sentiment_score = (polarity + 1) / 2 * 100
    
    # Store the timestamp and sentiment score
    timestamp = reply.created_at
    sentiment_scores.append((timestamp, sentiment_score))

sentiment_scores.sort(key=lambda x: x[0])

# Separate timestamps and scores for plotting
timestamps, scores = zip(*sentiment_scores)

# Create the plot
plt.figure(figsize=(10, 5))
plt.plot(timestamps, scores, marker='o')  # 'o' adds points at each data point
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d %H:%M'))
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))  # Adjust interval as needed
plt.xlabel('Time')
plt.ylabel('Sentiment Score (%)')
plt.title(f'Sentiment Analysis of Replies to Last 100 Posts by @{username}')
plt.grid(True)  # Optional: adds a grid for readability
plt.show()
```
